```python
from pyspark.sql.functions import *
```

```python
silver_df = spark.table("cricket_api.cricket.silver_current_matches")

silver_df.display()
```

```python
gold_match_type_df = silver_df.groupBy("match_type").agg(count("*").alias("match_count")).orderBy(desc("match_count"))
gold_match_type_df.display()
```

```python
gold_venue_df = silver_df.groupBy("venue").agg(count("*").alias("total_matches")).orderBy(desc("total_matches"))
gold_venue_df.display()
```

```python
#team-wise match count

team1_df = silver_df.select(col("team_1").alias("team"))

#team-wise match count

team2_df = silver_df.select(col("team_2").alias("team"))

all_team_df = team1_df.union(team2_df)

gold_team_df = all_team_df.groupBy("team").agg(count("*").alias("Teamwise_Total_Matches")).orderBy(desc("Teamwise_Total_Matches"))

gold_team_df.display()
```

```python
df_summary = spark.sql("""
SELECT 
    COUNT(*) AS Total_Matches,
    COUNT(DISTINCT match_type) AS Total_Match_Types,
    COUNT(DISTINCT venue) AS Total_Venues
FROM cricket_api.cricket.silver_current_matches
""")

df_summary.display()
```
